```python
# 2 - ANÁLISE EXPLORATÓRIA

import os
import mne
import pandas as pd
import numpy as np

proc_dir = "/workspaces/eeg-neuromodulation-mdd/data/processed"
arquivos_proc = [f for f in os.listdir(proc_dir) if f.endswith('.fif')]

# Dicionário de bandas para o TDM
bandas = {'Delta': (0.5, 4), 'Theta': (4, 8), 'Alpha': (8, 13), 'Beta': (13, 30)}

resultados = []

for arq in arquivos_proc:
    raw = mne.io.read_raw_fif(os.path.join(proc_dir, arq), preload=True)
    
    # Cálculo da PSD (Welch)
    psd = raw.compute_psd(method='welch', fmin=0.5, fmax=45)
    psd_data, freqs = psd.get_data(return_freqs=True)
    
    # Extrair potência média de cada banda
    dados_paciente = {'paciente': arq}
    for nome, (fmin, fmax) in bandas.items():
        # Filtro de frequências
        mask = (freqs >= fmin) & (freqs <= fmax)
        potencia = np.mean(psd_data[:, mask], axis=1)
        dados_paciente[nome] = np.mean(potencia) # Média de todos os canais
        
    resultados.append(dados_paciente)

# Criar DataFrame final para análise
df_features = pd.DataFrame(resultados)
print(df_features.head())
```

```python
# EXTRAÇÃO DAS FEATURES

import os
import pandas as pd
import mne
import numpy as np
from scipy import signal, stats

# --- 1. CONFIGURAÇÕES ---
proc_dir = "/workspaces/eeg-neuromodulation-mdd/data/processed"
meta_path = "/workspaces/eeg-neuromodulation-mdd/docs/metadados_template.csv"

# --- 2. EXTRAÇÃO DAS FEATURES ---
arquivos_proc = [f for f in os.listdir(proc_dir) if f.endswith('.fif')]
resultados = []

for arq in arquivos_proc:
    raw = mne.io.read_raw_fif(os.path.join(proc_dir, arq), preload=True)
    psd = raw.compute_psd(method='welch', fmin=0.5, fmax=45)
    psd_data, freqs = psd.get_data(return_freqs=True)
    
    # Cálculos das potências
    def get_pow(ch_names, fmin, fmax):
        indices = [raw.ch_names.index(ch) for ch in ch_names]
        f_idx = (freqs >= fmin) & (freqs <= fmax)
        return np.mean(psd_data[indices, :][:, f_idx])

    # Métricas
    afa = np.log(get_pow(['FP2', 'F4', 'F8'], 8, 13)) - np.log(get_pow(['FP1', 'F3', 'F7'], 8, 13))
    tbr = get_pow(raw.ch_names, 4, 8) / get_pow(raw.ch_names, 13, 30)
    
    # Coerência F3-F4
    f3, f4 = raw.get_data(picks=['F3'])[0], raw.get_data(picks=['F4'])[0]
    freqs_coh, Cxy = signal.coherence(f3, f4, fs=512, nperseg=1024)
    coerencia = np.mean(Cxy[(freqs_coh >= 13) & (freqs_coh <= 30)])
    
    # Entropia
    data = raw.get_data()
    ent = np.mean([stats.entropy(np.abs(data[i, :]) + 1e-12) for i in range(data.shape[0])])
    
    resultados.append({'paciente': arq, 'AFA': afa, 'TBR': tbr, 'Coerencia': coerencia, 'Entropia': ent})

df_features = pd.DataFrame(resultados)

# --- 3. MERGE FORÇADO (Sem tratamentos de string complexos) ---
df_meta = pd.read_csv(meta_path)

# Criamos a chave base apenas removendo o sufixo fixo que sabemos que existe
df_features['match_key'] = df_features['paciente'].str.replace('_cleaned.fif', '.csv')
df_meta['match_key'] = df_meta['nome_arquivo']

# Realiza o merge
df_final = pd.merge(df_features, df_meta, on='match_key', how='inner')

# --- 4. EXPORTAÇÃO ---
if not df_final.empty:
    df_final.to_csv("dados_mestrado_finais.csv", index=False)
    print(f"Sucesso! {len(df_final)} linhas geradas.")
    print(df_final[['paciente', 'grupo', 'AFA', 'TBR']].head())
else:
    print("ERRO: O merge ainda está vazio.")
    print("Features (match_key):", df_features['match_key'].tolist())
    print("Meta (match_key):", df_meta['match_key'].tolist())
```
