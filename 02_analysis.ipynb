```python
# 2 - ANÁLISE EXPLORATÓRIA

import os
import mne
import pandas as pd
import numpy as np

proc_dir = "/workspaces/eeg-neuromodulation-mdd/data/processed"
arquivos_proc = [f for f in os.listdir(proc_dir) if f.endswith('.fif')]

# Dicionário de bandas para o TDM
bandas = {'Delta': (0.5, 4), 'Theta': (4, 8), 'Alpha': (8, 13), 'Beta': (13, 30)}

resultados = []

for arq in arquivos_proc:
    raw = mne.io.read_raw_fif(os.path.join(proc_dir, arq), preload=True)
    
    # Cálculo da PSD (Welch)
    psd = raw.compute_psd(method='welch', fmin=0.5, fmax=45)
    psd_data, freqs = psd.get_data(return_freqs=True)
    
    # Extrair potência média de cada banda
    dados_paciente = {'paciente': arq}
    for nome, (fmin, fmax) in bandas.items():
        # Filtro de frequências
        mask = (freqs >= fmin) & (freqs <= fmax)
        potencia = np.mean(psd_data[:, mask], axis=1)
        dados_paciente[nome] = np.mean(potencia) # Média de todos os canais
        
    resultados.append(dados_paciente)

# Criar DataFrame final para análise
df_features = pd.DataFrame(resultados)
print(df_features.head())
```

```python
# EXTRAÇÃO DAS FEATURES (com base na literatura)

import mne
import numpy as np
from mne.connectivity import spectral_connectivity_epochs
from scipy.stats import entropy

# Definindo grupos de canais para Assimetria e Coerência
frontais_esq = ['FP1', 'F3', 'F7']
frontais_dir = ['FP2', 'F4', 'F8']

resultados_profundos = []

for arq in arquivos_proc:
    raw = mne.io.read_raw_fif(os.path.join(proc_dir, arq), preload=True)
    
    # 1. Cálculo da PSD para AFA e TBR
    psd = raw.compute_psd(method='welch', fmin=0.5, fmax=45)
    psd_data, freqs = psd.get_data(return_freqs=True)
    
    # Helper para extrair potência de canal/banda
    def get_pow(ch_names, fmin, fmax):
        indices = [raw.ch_names.index(ch) for ch in ch_names]
        f_idx = (freqs >= fmin) & (freqs <= fmax)
        return np.mean(psd_data[indices, :][:, f_idx])

    # AFA (Assimetria Frontal Alpha): log(dir) - log(esq)
    # Valores positivos sugerem maior atividade Alpha no lado direito (menor atividade cerebral esq)
    afa = np.log(get_pow(frontais_dir, 8, 13)) - np.log(get_pow(frontais_esq, 8, 13))
    
    # TBR (Theta/Beta Ratio) - Média global
    tbr = get_pow(raw.ch_names, 4, 8) / get_pow(raw.ch_names, 13, 30)
    
    # 2. Coerência Frontal (F3-F4)
    # Convertemos o sinal para épocas (necessário para conectividade)
    epochs = mne.make_fixed_length_epochs(raw, duration=2, overlap=0.5)
    con = spectral_connectivity_epochs(epochs, method='coh', fmin=13, fmax=30, verbose=False)
    coerencia = np.mean(con.get_data(output='dense'))
    
    # 3. Entropia (Baseada na sinal temporal - Dimensão de Permutação)
    # Calculamos no sinal bruto para capturar a complexidade da dinâmica
    data = raw.get_data()
    ent = np.mean([entropy(data[i, :]) for i in range(data.shape[0])])
    
    resultados_profundos.append({
        'paciente': arq,
        'AFA': afa,
        'TBR': tbr,
        'Coerencia': coerencia,
        'Entropia': ent
    })

df_profundo = pd.DataFrame(resultados_profundos)
print(df_profundo.head())
```
